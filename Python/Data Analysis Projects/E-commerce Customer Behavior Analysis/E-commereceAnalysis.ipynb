# Accessing the missing values 

```python
filtered_df = order_reviews_df[order_reviews_df[['review_comment_title','review_comment_message']].isnull().any(axis=1)]

filtered_df[['review_comment_title','review_comment_message']]
```

```python
filtered_df = orders_df[orders_df[["order_approved_at", "order_delivered_carrier_date" , "order_delivered_customer_date"]].isnull().any(axis=1)]
filtered_df[["order_approved_at", "order_delivered_carrier_date" , "order_delivered_customer_date"]]
```

```python
filtered_df = products_df[products_df.isnull().any(axis=1)]
filtered_df
```

# FILLING THE MISSING REVIEW TEXT

```python
order_reviews_df['review_comment_title'] = order_reviews_df['review_comment_title'].fillna("No Title")
order_reviews_df['review_comment_message'] = order_reviews_df['review_comment_message'].fillna("No Message")
order_reviews_df.head()
```

# Convert date columns to datetime objects

```python
date_columns = ['order_purchase_timestamp',	'order_approved_at',	
                'order_delivered_carrier_date',	'order_delivered_customer_date','order_estimated_delivery_date'
               ]

for cols in date_columns:
    orders_df[cols] = pd.to_datetime(orders_df[cols])

orders_df.dtypes    
```

# Handle Product Category NaNs

```python
products_df['product_category_name'] = products_df['product_category_name'].fillna("Unknown")
products_df.head()
```
