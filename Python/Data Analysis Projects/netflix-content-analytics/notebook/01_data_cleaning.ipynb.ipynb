# 🧼 Data Cleaning & Feature Engineering

## 📌 Objective
The goal of this notebook is to clean the raw Netflix dataset and prepare it for further analysis by handling data types, missing values, and extracting useful features.

---

## 📂 Dataset Overview
The dataset contains information about Netflix Originals including:
- Title
- Genre
- Premiere Date
- Runtime
- IMDB Score
- Language

```python
import pandas as pd

df = pd.read_csv("../data/netflix_content_dataset.csv", encoding="ISO-8859-1")

df.head()
df.info()
df.isnull().sum()
```

## 🔍 Initial Data Inspection

- Checked dataset structure using `df.info()`
- Verified missing values using `df.isnull().sum()`
- Observed that:
  - No missing values in key columns
  - 'Premiere' column is in object format
  - 'Runtime' needs cleaning

```python
# Fix date
df['Premiere'] = pd.to_datetime(df['Premiere'], errors='coerce')
df = df.dropna(subset=['Premiere'])

# Extract features
df['release_year'] = df['Premiere'].dt.year
df['month'] = df['Premiere'].dt.month_name()

# Clean runtime safely
df['duration_min'] = (
    df['Runtime']
    .astype(str)
    .str.replace(' min','', regex=False)
    .str.replace('min','', regex=False)
)

df['duration_min'] = pd.to_numeric(df['duration_min'], errors='coerce')
df = df.dropna(subset=['duration_min'])

print("Dataset Shape:", df.shape)
df.head()
```

## ⚙️ Feature Engineering

To make the dataset more useful for analysis, the following transformations were applied:

### ✔ Date Conversion
- Converted 'Premiere' column to datetime format
- Handled invalid values using `errors='coerce'`

### ✔ New Features Created
- `release_year` → Extracted year from premiere date
- `month` → Extracted month name

### ✔ Runtime Cleaning
- Removed 'min' text from runtime column
- Converted runtime into numeric format (`duration_min`)

### 📏 Dataset Size
The dataset contains rows and columns after cleaning, ensuring data is ready for analysis.

### 🔍 Insight:
After cleaning, the dataset contains 579 records and 9 features, ensuring high data quality for analysis.
